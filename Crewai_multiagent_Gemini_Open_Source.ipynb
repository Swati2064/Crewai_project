```python
import os
from crewai import Agent, Task, Crew, LLM
from crewai_tools import SerperDevTool

# ============================================================
# 1. API KEYS
# ============================================================

os.environ["GEMINI_API_KEY"] = ""
os.environ["SERPER_API_KEY"] = ""


# ============================================================
# 2. GEMINI LLM
# ============================================================

llm = LLM(
    model="gemini/gemini-2.5-flash",
    api_key=os.environ["GEMINI_API_KEY"],
    temperature=0.5
)


# ============================================================
# 3. SEARCH TOOL
# ============================================================

search_tool = SerperDevTool()


# ============================================================
# 4. AGENTS
# ============================================================

teacher = Agent(
    role="Teacher",
    goal="Explain concepts clearly step by step",
    backstory="Experienced teacher who explains concepts with simple examples",
    llm=llm,
    verbose=True
)

researcher = Agent(
    role="Researcher",
    goal="Find important and useful information",
    backstory="Expert researcher who finds relevant real-world information",
    tools=[search_tool],
    llm=llm,
    verbose=True
)

simplifier = Agent(
    role="Simplifier",
    goal="Make complex concepts very easy to understand",
    backstory="Expert at breaking complex technical concepts into simple language",
    llm=llm,
    verbose=True
)

student = Agent(
    role="Student",
    goal="Create simple study notes",
    backstory="A student who writes short and easy-to-understand notes",
    llm=llm,
    verbose=True
)

examiner = Agent(
    role="Examiner",
    goal="Create questions to test understanding",
    backstory="Experienced examiner who creates useful questions",
    llm=llm,
    verbose=True
)


# ============================================================
# 5. TOPIC
# ============================================================

topic = """
What is the difference between Agents, Agentic AI,
MCP (Model Context Protocol), and Generative AI?
"""


# ============================================================
# 6. TASKS
# ============================================================

task1 = Task(
    description=f"""
    Explain the following topic in simple steps:

    {topic}

    Give simple examples for each concept.
    """,
    expected_output="A clear step-by-step explanation with examples.",
    agent=teacher
)

task2 = Task(
    description=f"""
    Research the following topic:

    {topic}

    Find 3 important points that help explain the differences.
    """,
    expected_output="Three clear and useful points.",
    agent=researcher
)

task3 = Task(
    description=f"""
    Simplify this topic for a beginner:

    {topic}

    Use very simple English and easy examples.
    """,
    expected_output="A very simple beginner-friendly explanation.",
    agent=simplifier
)

task4 = Task(
    description=f"""
    Create short student notes about:

    {topic}

    Keep the notes simple and easy to remember.
    """,
    expected_output="Short and simple study notes.",
    agent=student
)

task5 = Task(
    description=f"""
    Create 5 questions to test a student's understanding of:

    {topic}

    Include easy questions.
    """,
    expected_output="Five easy questions about the topic.",
    agent=examiner
)


# ============================================================
# 7. CREATE CREW
# ============================================================

crew = Crew(
    agents=[
        teacher,
        researcher,
        simplifier,
        student,
        examiner
    ],
    tasks=[
        task1,
        task2,
        task3,
        task4,
        task5
    ],
    verbose=True
)


# ============================================================
# 8. RUN CREW
# ============================================================

result = await crew.kickoff_async()

print("\n\n================ FINAL RESULT ================\n")
print(result)
```
